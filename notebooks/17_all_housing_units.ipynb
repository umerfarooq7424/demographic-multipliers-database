```python
import pandas as pd
import numpy as np

# Define a function to calculate confidence intervals and error margins
def calculate_confidence_interval(row, column_names,mean):
    z_score = 1.645  # Z-score for 90% confidence
    
    # Lists to collect values for each age group
    means = []
    standard_errors = []
    lower_bounds = []
    upper_bounds = []
    error_margins = []
    
    # For each age group, calculate the mean, SE, CI, and Error Margin
    for age_group in column_names:
        
        try:
            # Ensure mean_value is a float
            mean_value = float(row[age_group])
        except (ValueError, TypeError):
            # If conversion fails, skip this row
            continue
        
        total_persons = float(row[mean]) if row[mean] != '' else 0
        if total_persons == 0:
            se = np.nan
        else:
            se = mean_value / np.sqrt(total_persons)
        
        # Calculate confidence intervals
        lower_bound = mean_value - z_score * se
        upper_bound = mean_value + z_score * se
        
        # Calculate error margin as percentage
        error_margin_pct = ((upper_bound - lower_bound) / (2 * mean_value)) *100  if mean_value != 0 else np.nan
        
        # Append values to the lists
        means.append(mean_value)
        standard_errors.append(se)
        lower_bounds.append(lower_bound)
        upper_bounds.append(upper_bound)
        error_margins.append(error_margin_pct)
    
    # Calculate the average values across all age groups
    if len(means) > 0:
        overall_mean = np.nanmean(means)
        overall_se = np.nanmean(standard_errors)
        overall_lower = np.nanmean(lower_bounds)
        overall_upper = np.nanmean(upper_bounds)
        overall_error_margin_pct = np.nanmean(error_margins)
    else:
        # Handle the case where there are no valid values
        overall_mean = overall_se = overall_lower = overall_upper = overall_error_margin_pct = np.nan
    
    # Return the combined results as a single row
    average = float(row[mean]) if row[mean] != '' else 0
     
    if average == 0:
        return pd.Series({
            'Standard Errors': '',
            'Low': '',
            'High': '',
            'Error Margin as %': ''
        })
    else:
        return pd.Series({
            'Standard Errors': overall_se,
            'Low': overall_lower + average,
            'High': overall_upper + average,
            'Error Margin as %': ((overall_upper + average) - (overall_lower + average)) / (2 * average) * 100
        })

# Define your age groups list
age_groups_list = ['0-4', '5-17', '18-34', '35-44', '45-54', '55-64', '65-74', '75+']

# Apply the function to each row
confidence_intervals_df = df4.apply(calculate_confidence_interval, axis=1, column_names=age_groups_list,mean='TOTAL PERSONS')

# Combine the new columns with the original DataFrame
df4 = pd.concat([df4, confidence_intervals_df], axis=1)

# Fill any missing values if needed
df4 = df4.fillna('')
df4 = df4.drop(columns=age_groups_list)  # Drop the original age group columns
# Display the updated DataFrame
df4.head(10)
```
