```python
import pandas as pd
import numpy as np
from glob import glob
import os
```

## 1. Demographic Multipliers (DMs)

### Population by age

```python
def proc_DM_tbII_1(state_abbr, unit_type = 'ALL'):

    sheet_nm = 'TABLE IIB1'
    if unit_type == 'NEWER':
        sheet_nm = 'TABLE IIA1'
    raw_DM_all_NJ = pd.read_excel(
        io = glob(f'data_raw/*{state_abbr}*{unit_type}*xlsx')[0], 
        sheet_name = sheet_nm,
        skiprows = 4,
    )

    raw_DM_all_NJ = raw_DM_all_NJ[~raw_DM_all_NJ['VALUE/TENURE'].isna()].iloc[:,1:]
    raw_DM_all_NJ_structure = raw_DM_all_NJ.iloc[:,0].copy()
    
    raw_DM_all_NJ_structure_idx = raw_DM_all_NJ_structure.value_counts()
    raw_DM_all_NJ_structure_idx = raw_DM_all_NJ_structure_idx[raw_DM_all_NJ_structure_idx>1].index.tolist()
    
    raw_DM_all_NJ_structure.loc[raw_DM_all_NJ_structure.isin(raw_DM_all_NJ_structure_idx)] = ''
    
    for i, txt in raw_DM_all_NJ_structure.items():
        if txt == '':
            raw_DM_all_NJ_structure[i] = raw_DM_all_NJ_structure[i-1]
    
    raw_DM_all_NJ['Structure']=raw_DM_all_NJ_structure
    raw_DM_all_NJ = raw_DM_all_NJ[~raw_DM_all_NJ.PERSONS.isna()]
    
    del raw_DM_all_NJ_structure, raw_DM_all_NJ_structure_idx
    
    raw_DM_all_NJ['value_range'] = raw_DM_all_NJ.apply(
        lambda x: f'{x[1]} {x[2]} {x[3]}' if unit_type == 'ALL' else f'{x[1]} {x[2]}', 
        axis= 1
    ).str.replace(' *nan *|\\.0','', regex = True)
    
    raw_DM_all_NJ = raw_DM_all_NJ[['Structure','VALUE/TENURE','value_range',
                  'PERSONS',
                  '0-4', '5-17', '18-34', '35-44', '45-54', '55-64', '65-74', '75+',]]
    raw_DM_all_NJ = raw_DM_all_NJ.rename({'VALUE/TENURE':'VALUE_TENURE'}, axis=1)
    raw_DM_all_NJ = raw_DM_all_NJ.reset_index(drop = True)
    return raw_DM_all_NJ
```

### School age children (SAC) by grade

```python
def proc_DM_tbII_2(state_abbr, unit_type = 'ALL'):

    sheet_nm = 'TABLE IIB2'
    if unit_type == 'NEWER':
        sheet_nm = 'TABLE IIA2'

    raw_DM_all_NJ = pd.read_excel(
        io = glob(f'data_raw/*{state_abbr}*{unit_type}*xlsx')[0], 
        sheet_name = sheet_nm,
        skiprows = 5,
    )

    raw_DM_all_NJ = raw_DM_all_NJ[~raw_DM_all_NJ['VALUE/TENURE'].isna()].iloc[:,1:]
    raw_DM_all_NJ_structure = raw_DM_all_NJ.iloc[:,0].copy()
    
    raw_DM_all_NJ_structure_idx = raw_DM_all_NJ_structure.value_counts()
    raw_DM_all_NJ_structure_idx = raw_DM_all_NJ_structure_idx[raw_DM_all_NJ_structure_idx>1].index.tolist()
    
    raw_DM_all_NJ_structure.loc[raw_DM_all_NJ_structure.isin(raw_DM_all_NJ_structure_idx)] = ''
    
    for i, txt in raw_DM_all_NJ_structure.items():
        if txt == '':
            raw_DM_all_NJ_structure[i] = raw_DM_all_NJ_structure[i-1]
    
    raw_DM_all_NJ['Structure']=raw_DM_all_NJ_structure
    raw_DM_all_NJ = raw_DM_all_NJ[~raw_DM_all_NJ.SAC.isna()]

    del raw_DM_all_NJ_structure, raw_DM_all_NJ_structure_idx
    
    raw_DM_all_NJ['value_range'] = raw_DM_all_NJ.apply(
        lambda x: f'{x[2]} {x[3]} {x[4]}' if unit_type == 'ALL' else f'{x[2]}', 
        axis= 1
    ).str.replace(' *nan *|\\.0','', regex = True)

    raw_DM_all_NJ = raw_DM_all_NJ[['Structure','VALUE/TENURE','value_range',
                  'SAC',
                  '(K-5)', '(6-8)', '(9-12)',]]
    raw_DM_all_NJ = raw_DM_all_NJ.rename({'VALUE/TENURE':'VALUE_TENURE'},axis=1)
    raw_DM_all_NJ = raw_DM_all_NJ.reset_index(drop = True)
    return raw_DM_all_NJ
```
